# SQL

```python
df = pd.read_csv('nhs_ae_clean.csv')
```

```python
df.head()
```

```python
import sqlite3
```

```python
#create database in memory 

conn = sqlite3.connect(':memory:')

```

```python
#load out clean data into it

df.to_sql('nhs_ae', conn, index=False, if_exists='replace')
```

```python
print("Table name: nhs_ae")
```

```python
query = """
select 
    count(*) as total_months,
    round(avg(pct_4h_type1)*100, 1) as avg_performance_pct,
    round(min(pct_4h_type1)*100, 1) as worst_ever_pct,
    round(max(pct_4h_type1)*100, 1) as best_ever_pct
from nhs_ae
"""
```

```python
result = pd.read_sql(query, conn)

print(result)
```

```python
# Find WHICH year was worst
```

```python
query = """

select
    strftime('%Y', period) as year,
    round(avg(pct_4h_type1)*100, 1) as avg_performance_pct,
    count(*) as months
from nhs_ae
group by year
order by avg_performance_pct asc
"""
```

```python
result = pd.read_sql(query, conn)
print(result)
```

```python
# Find which MONTH is worst every year
```

```python
query = """
SELECT 
    strftime('%m', period) as month_num,
    strftime('%B', period) as month_name,
    ROUND(AVG(pct_4h_type1) * 100, 1) as avg_performance_pct
FROM nhs_ae
GROUP BY month_num
ORDER BY avg_performance_pct ASC
"""
```

```python
result = pd.read_sql(query, conn)
print(result)
```

```python
# Count how many months breached the 95% target

query = """
SELECT 
    COUNT(*) as total_months,
    SUM(CASE WHEN pct_4h_type1 < 0.95 THEN 1 ELSE 0 END) as months_in_breach,
    SUM(CASE WHEN pct_4h_type1 >= 0.95 THEN 1 ELSE 0 END) as months_on_target
FROM nhs_ae
"""

result = pd.read_sql(query, conn)
print(result)
```

# Visulisation

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
```

```python
# 15 Years of Performance

plt.figure(figsize=(14, 6))

plt.plot(df['period'], df['pct_4h_type1'] * 100, 
         color='steelblue', linewidth=2, label='Actual Performance')

plt.axhline(y=95, color='red', linestyle='--', 
            linewidth=2, label='NHS 95% Target')

plt.title('NHS A&E 4-Hour Performance: 2010 to 2026', fontsize=16)
plt.xlabel('Year')
plt.ylabel('% Patients Seen Within 4 Hours')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Yearly Performance Bar Chart

query = """
SELECT 
    strftime('%Y', period) as year,
    ROUND(AVG(pct_4h_type1) * 100, 1) as avg_performance_pct
FROM nhs_ae
GROUP BY year
ORDER BY year ASC
"""
df_yearly = pd.read_sql(query, conn)

plt.figure(figsize=(14, 6))

colors = ['red' if x < 95 else 'green' for x in df_yearly['avg_performance_pct']]

plt.bar(df_yearly['year'], df_yearly['avg_performance_pct'], color=colors)

plt.axhline(y=95, color='black', linestyle='--', linewidth=2, label='95% Target')

plt.title('NHS A&E Average Performance by Year', fontsize=16)
plt.xlabel('Year')
plt.ylabel('% Patients Seen Within 4 Hours')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Seasonality Chart: Winter vs Summer

query = """
SELECT 
    strftime('%m', period) as month_num,
    ROUND(AVG(pct_4h_type1) * 100, 1) as avg_performance_pct
FROM nhs_ae
GROUP BY month_num
ORDER BY month_num ASC
"""
df_monthly = pd.read_sql(query, conn)

df_monthly['month_name'] = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                             'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

colors = ['steelblue' if x >= 79 else 'tomato' for x in df_monthly['avg_performance_pct']]

plt.figure(figsize=(12, 6))

plt.bar(df_monthly['month_name'], df_monthly['avg_performance_pct'], color=colors)

plt.axhline(y=95, color='black', linestyle='--', linewidth=2, label='95% Target')

plt.title('NHS A&E Performance by Month (Seasonality)', fontsize=16)
plt.xlabel('Month')
plt.ylabel('% Patients Seen Within 4 Hours')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Breach Risk Chart: How Close to Target Each Year

query = """
SELECT 
    strftime('%Y', period) as year,
    ROUND(AVG(pct_4h_type1) * 100, 1) as avg_performance_pct,
    ROUND((95 - AVG(pct_4h_type1) * 100), 1) as gap_from_target
FROM nhs_ae
GROUP BY year
ORDER BY year ASC
"""
df_gap = pd.read_sql(query, conn)

plt.figure(figsize=(14, 6))

colors = ['red' if x > 0 else 'green' for x in df_gap['gap_from_target']]

plt.bar(df_gap['year'], df_gap['gap_from_target'], color=colors)

plt.axhline(y=0, color='black', linestyle='--', linewidth=2, label='95% Target Line')

plt.title('NHS A&E Gap From 95% Target by Year', fontsize=16)
plt.xlabel('Year')
plt.ylabel('% Points Below Target')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```
